```python
## 1. Define Base Paths using python variable
base_path = "/Volumes/workspace/default/volumewd36"

raw_path = f"{base_path}/raw"
processed_path = f"{base_path}/processed"
archive_path = f"{base_path}/archive"
```

```python
## 2. dbutils Usecase – Create Directories using the above path variables..

dbutils.fs.mkdirs(raw_path)
dbutils.fs.mkdirs(processed_path)
dbutils.fs.mkdirs(archive_path)
```

```python
## 3. dbutils Usecase – Create Sample Healthcare File
sample_data = """patient_id,patient_name,age,gender
1,John Doe,68,M
2,Jane Smith,54,F
"""
dbutils.fs.put(f"{raw_path}/patients.csv", sample_data, overwrite=True)
```

```python
## 4. dbutils Usecase - list the file created

display(dbutils.fs.ls(raw_path))
```

```python
## 6. dbutils widget usecase - Create dropdown and text widgets...

dbutils.widgets.dropdown("environment", "dev", ["dev", "qa", "prod"], "Environment")
dbutils.widgets.text("owner", "", "Owner Name")
```

```python
## 7. dbutils widget Usecase – Read Widget Values environment and owner and print in the screen

environment = dbutils.widgets.get("environment")
owner = dbutils.widgets.get("owner")

print(f"Environment : {environment}")
print(f"Owner       : {owner}")
```

```python
## 8. dbutils widget Usecase – Move the above processed File to Archive
print(f"Processed Paht :{processed_path}")
print(f"Archive Paht :{archive_path}")
display(dbutils.fs.ls(base_path))

dbutils.fs.cp(f"{raw_path}/patients.csv", f"{processed_path}/patients.csv")

dbutils.fs.mv(f"{processed_path}/patients.csv", f"{archive_path}/patients.csv")
```
