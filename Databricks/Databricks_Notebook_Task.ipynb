```python
%py
import requests

url = "https://public.tableau.com/app/sample-data/mobile_os_usage.csv"

# 1. API call: pull the data into a Python variable
response = requests.get(url, timeout=30)
response.raise_for_status()   # stops with a clear error if the call failed (404, 403, etc.)

# 2. Write the data into the volume using the variable's text
dbutils.fs.put(
    "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv",
    response.text,
    overwrite=True
)
```
