```python
%sql
CREATE TABLE IF NOT EXISTS  default.customers (
  CustomerID   INT       NOT NULL,
  FirstName    STRING,
  LastName     STRING,
  DateOfBirth  DATE,
  Email        STRING,
  PhoneNumber  STRING,
  Address      STRING,
  BranchID     INT
)
```

```python
%sql
COPY INTO customers
FROM '/Volumes/workspace/default/datalake/dbfs_data'
FILEFORMAT = CSV
PATTERN = 'cust*csv'
FORMAT_OPTIONS ('header' = 'true', 'inferSchema' = 'true', 'dateFormat' = 'yyyy-MM-dd')
COPY_OPTIONS ('mergeSchema' = 'true');
```

```python
%sql
DROP TABLE IF EXISTS default.employees;

CREATE TABLE IF NOT EXISTS default.employees (
  EmployeeID  INT  NOT NULL,
  FirstName   STRING,
  LastName    STRING,
  Position    STRING,
  HireDate    DATE,
  ManagerID   BIGINT,
  BranchID    INT
)

```

```python
%sql
ALTER TABLE employees ALTER COLUMN EmployeeID DROP NOT NULL;

INSERT INTO employees (EmployeeID, FirstName, LastName, Position, HireDate, ManagerID, BranchID)
SELECT
  try_cast(EmployeeID AS INT),
  FirstName,
  LastName,
  Position,
  to_date(Hiredate, 'dd-MM-yyyy'),
  try_cast(ManagerID AS BIGINT),
  try_cast(BranchID AS INT)
FROM read_files(
  '/Volumes/workspace/default/datalake/dbfs_data/employee.csv',
  format => 'csv',
  header => true,
  schema => 'BranchID STRING, EmployeeID STRING, FirstName STRING, Hiredate STRING, LastName STRING, ManagerID STRING, Position STRING'
)
```
